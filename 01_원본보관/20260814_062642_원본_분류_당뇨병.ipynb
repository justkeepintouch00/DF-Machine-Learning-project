#### 추가 피드백 후 개선점 1번  
이상치(Outlier) 처리

`BMI`, `MentHlth`, `PhysHlth`와 같이 값의 범위가 넓은 연속형 변수에서 IQR(Interquartile Range) 방법을 사용하여 이상치를 식별하고 제거합니다. 이상치는 모델 학습에 부정적인 영향을 줄 수 있습니다.

```python
df_processed = df_cleaned.copy()

# IQR을 이용한 이상치 제거 함수
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    initial_count = df.shape[0]
    df_filtered = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    removed_count = initial_count - df_filtered.shape[0]
    print(f"'{column}' 컬럼에서 제거된 이상치: {removed_count}개")
    return df_filtered

# 선택된 컬럼에 이상치 제거 적용
columns_to_check = ['BMI', 'MentHlth', 'PhysHlth']
for col in columns_to_check:
    df_processed = remove_outliers_iqr(df_processed, col)

print(f"\n최종 데이터 행 수 (중복 및 이상치 제거 후): {df_processed.shape[0]}")
```

### 추가 피드백 후 개선점 3번
3. 다양한 계열의 앙상블 모델 활용 (LightGBM + Logistic Regression)

이전 개선점에서 SMOTETomek과 LightGBM 모델의 높은 재현율을 확인했습니다. 이를 Logistic Regression (선형 계열)과 결합한 앙상블 모델을 구성하여, 두 모델의 강점을 활용하고 예측 오류를 줄이는 방법을 시도합니다.

`VotingClassifier`를 사용하여 두 모델의 예측 확률을 종합하고, 최종 예측값을 도출합니다. 앙상블 구성 시, LightGBM 모델과 Logistic Regression 모델 모두 SMOTETomek으로 재샘플링된 데이터를 사용하여 훈련됩니다. 이는 클래스 불균형에 대응하고 모델의 일반화 성능을 향상시키는 데 기여합니다.

```python
from imblearn.combine import SMOTETomek
import lightgbm as lgb
from sklearn.linear_model import LogisticRegression # 로지스틱 회귀 모델을 위한 임포트
from sklearn.ensemble import VotingClassifier # 앙상블 모델을 위한 임포트
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

def get_clf_eval(y_test, y_pred, y_pred_proba=None):
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_pred_proba) if y_pred_proba is not None else 'N/A'

    print(f'정확도: {accuracy:.4f}')
    print(f'정밀도: {precision:.4f}')
    print(f'재현율: {recall:.4f}')
    print(f'F1 스코어: {f1:.4f}')
    print(f'ROC AUC: {roc_auc:.4f}')


print('\n===== SMOTETomek 적용 및 LightGBM + Logistic Regression 앙상블 모델 훈련 및 평가 =====')

# SMOTETomek 인스턴스 생성
smote_tomek = SMOTETomek(random_state=42)

# 훈련 데이터에 SMOTETomek 적용
X_train_resampled, Y_train_resampled = smote_tomek.fit_resample(X_train_proc, Y_train_proc)

print(f"SMOTETomek 적용 전 훈련 데이터 클래스 분포:\n{Y_train_proc.value_counts()}")
print(f"SMOTETomek 적용 후 훈련 데이터 클래스 분포:\n{Y_train_resampled.value_counts()}")

# 1. LightGBM 모델 훈련 (resampled 데이터 사용)
lgbm_smote_tomek = lgb.LGBMClassifier(objective='binary', random_state=42)
lgbm_smote_tomek.fit(X_train_resampled, Y_train_resampled)
print('\nLightGBM (SMOTETomek) 개별 모델 훈련 완료.')

# LightGBM 개별 모델 예측 결과 (하위 셀 호환성을 위해 유지)
Y_pred_lgbm_st = lgbm_smote_tomek.predict(X_test_proc)
Y_probs_lgbm_st = lgbm_smote_tomek.predict_proba(X_test_proc)[:, 1]

# 2. Logistic Regression 모델 훈련 (resampled 데이터 사용)
lr_clf = LogisticRegression(solver='liblinear', class_weight='balanced', random_state=42)
lr_clf.fit(X_train_resampled, Y_train_resampled)
print('\nLogistic Regression (SMOTETomek) 개별 모델 훈련 완료.')

# 3. VotingClassifier 정의 및 훈련 (soft voting으로 확률 평균 사용)
voting_clf = VotingClassifier(
    estimators=[('lgbm_st', lgbm_smote_tomek), ('lr_st', lr_clf)],
    voting='soft',
    n_jobs=-1
)

# 앙상블 모델 훈련 (SMOTETomek으로 재샘플링된 데이터 사용)
voting_clf.fit(X_train_resampled, Y_train_resampled)
print('\n앙상블 (LightGBM + Logistic Regression with SMOTETomek) 모델 훈련 완료.')

# 테스트 데이터로 앙상블 예측 및 평가
Y_pred_ensemble = voting_clf.predict(X_test_proc)
Y_probs_ensemble = voting_clf.predict_proba(X_test_proc)[:, 1]

print('\n앙상블 (LightGBM (SMOTETomek) + Logistic Regression (SMOTETomek)) 모델 평가:')
get_clf_eval(Y_test_proc, Y_pred_ensemble, Y_probs_ensemble)
```

### SMOTETomek + 앙상블 모델 여러 시드로 반복 훈련 및 평균 평가

모델의 안정성과 일반화 성능을 확인하기 위해 여러 개의 `random_state`를 사용하여 SMOTETomek 재샘플링 및 앙상블 모델(LightGBM + Logistic Regression) 훈련 과정을 반복하고, 각 실행의 평가 지표(정확도, 정밀도, 재현율, F1 스코어, ROC AUC)를 수집하여 최종 평균 성능을 계산합니다. 이는 특정 `random_state`에 과적합되는 것을 방지하고 모델의 견고성을 평가하는 데 도움을 줍니다.

```python
import numpy as np
import lightgbm as lgb
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import VotingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from imblearn.combine import SMOTETomek

def train_and_evaluate_ensemble_with_smotetomek(X_train, Y_train, X_test, Y_test, random_seed):
    """
    주어진 random_seed로 SMOTETomek을 적용하고,
    LightGBM과 Logistic Regression 앙상블 모델을 훈련 및 평가하는 함수.
    """
    print(f"\n--- Running with random_state = {random_seed} ---")

    # SMOTETomek 인스턴스 생성 및 적용
    smote_tomek = SMOTETomek(random_state=random_seed)
    X_train_resampled, Y_train_resampled = smote_tomek.fit_resample(X_train, Y_train)

    # LightGBM 모델 훈련
    lgbm_smote_tomek = lgb.LGBMClassifier(objective='binary', random_state=random_seed)
    lgbm_smote_tomek.fit(X_train_resampled, Y_train_resampled)

    # Logistic Regression 모델 훈련
    lr_clf = LogisticRegression(solver='liblinear', class_weight='balanced', random_state=random_seed)
    lr_clf.fit(X_train_resampled, Y_train_resampled)

    # VotingClassifier 정의 및 훈련
    voting_clf = VotingClassifier(
        estimators=[('lgbm_st', lgbm_smote_tomek), ('lr_st', lr_clf)],
        voting='soft',
        n_jobs=-1
    )
    voting_clf.fit(X_train_resampled, Y_train_resampled)

    # 테스트 데이터로 예측 및 평가
    Y_pred_ensemble = voting_clf.predict(X_test)
    Y_probs_ensemble = voting_clf.predict_proba(X_test)[:, 1]

    accuracy = accuracy_score(Y_test, Y_pred_ensemble)
    precision = precision_score(Y_test, Y_pred_ensemble, zero_division=0)
    recall = recall_score(Y_test, Y_pred_ensemble)
    f1 = f1_score(Y_test, Y_pred_ensemble)
    roc_auc = roc_auc_score(Y_test, Y_probs_ensemble)

    print(f'정확도: {accuracy:.4f}, 정밀도: {precision:.4f}, 재현율: {recall:.4f}, F1: {f1:.4f}, ROC AUC: {roc_auc:.4f}')
    return accuracy, precision, recall, f1, roc_auc

# 여러 random_state 값으로 테스트
seeds = [42, 123, 789, 500, 999]
all_accuracies, all_precisions, all_recalls, all_f1s, all_roc_aucs = [], [], [], [], []

for seed in seeds:
    acc, prec, rec, f1, roc_auc = train_and_evaluate_ensemble_with_smotetomek(
        X_train_proc, Y_train_proc, X_test_proc, Y_test_proc, seed
    )
    all_accuracies.append(acc)
    all_precisions.append(prec)
    all_recalls.append(rec)
    all_f1s.append(f1)
    all_roc_aucs.append(roc_auc)

# 평균 성능 계산 및 출력
print("\n===== 평균 성능 결과 (여러 시드) =====")
print(f"평균 정확도: {np.mean(all_accuracies):.4f}")
print(f"평균 정밀도: {np.mean(all_precisions):.4f}")
print(f"평균 재현율: {np.mean(all_recalls):.4f}")
print(f"평균 F1 스코어: {np.mean(all_f1s):.4f}")
print(f"평균 ROC AUC: {np.mean(all_roc_aucs):.4f}")
```

### 오차 행렬 해석 및 예측 오류 유형

위 오차 행렬은 GridSearchCV를 통해 최적화된 RandomForest 모델의 테스트 데이터 예측 결과를 보여줍니다.

*   **True Negatives (TN):** 실제 0 (당뇨병 아님)인 것을 모델이 0 (당뇨병 아님)으로 정확히 예측한 경우의 수
*   **False Positives (FP):** 실제 0 (당뇨병 아님)인 것을 모델이 1 (당뇨병)으로 잘못 예측한 경우의 수 **(1종 오류)**
*   **False Negatives (FN):** 실제 1 (당뇨병)인 것을 모델이 0 (당뇨병 아님)으로 잘못 예측한 경우의 수 **(2종 오류)**
*   **True Positives (TP):** 실제 1 (당뇨병)인 것을 모델이 1 (당뇨병)으로 정확히 예측한 경우의 수

**주요 예측 오류:**

1.  **False Positives (FP) - 1종 오류:**
    *   모델이 당뇨병이 아닌 사람을 당뇨병이라고 잘못 예측하는 경우입니다. 이 경우 환자에게 불필요한 추가 검사나 심리적 부담을 줄 수 있습니다.

2.  **False Negatives (FN) - 2종 오류:**
    *   모델이 실제 당뇨병 환자를 당뇨병이 아니라고 잘못 예측하는 경우입니다. 이는 당뇨병 진단 모델에서 더 위험할 수 있는 오류입니다. 실제 질병이 있는데도 놓치게 되면 적절한 시기에 치료를 받지 못해 질병이 악화될 수 있습니다.

오차 행렬을 통해 어떤 유형의 오류가 더 많이 발생하는지 파악하고, 모델의 목적에 따라 (예: 2종 오류를 줄이는 것이 더 중요한 경우) 모델 개선 전략을 세울 수 있습니다.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np

# 최적 모델로부터 예측 확률 얻기
Y_probs = grid_search.best_estimator_.predict_proba(X_test)[:, 1] # 양성 클래스(1)에 대한 확률

thresholds = np.arange(0.1, 1.0, 0.05) # 0.1부터 0.95까지 0.05 간격으로 임계값 설정

accuracy_scores = []
precision_scores = []
recall_scores = []
f1_scores = []

for threshold in thresholds:
    Y_pred_thresh = (Y_probs >= threshold).astype(int)

    accuracy_scores.append(accuracy_score(Y_test, Y_pred_thresh))
    precision_scores.append(precision_score(Y_test, Y_pred_thresh, zero_division=0))
    recall_scores.append(recall_score(Y_test, Y_pred_thresh))
    f1_scores.append(f1_score(Y_test, Y_pred_thresh))

plt.figure(figsize=(12, 6))
plt.plot(thresholds, accuracy_scores, label='Accuracy', marker='o')
plt.plot(thresholds, precision_scores, label='Precision', marker='o')
plt.plot(thresholds, recall_scores, label='Recall', marker='o')
plt.plot(thresholds, f1_scores, label='F1 Score', marker='o')

plt.title('Evaluation Metrics vs. Classification Threshold')
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.legend()
plt.grid(True)
plt.xticks(thresholds)
plt.show()
```
